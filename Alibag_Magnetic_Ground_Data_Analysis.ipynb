# Alibag Ground Magnetic Data from 16 to 26 January 2026

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

Reading the file and coverting date and time to datetime

```python
dir = r'C:\Users\eleen\OneDrive\Desktop\Event_Based_Project\16_26_alibag_magnetic data.min'

df = pd.read_csv(dir,skiprows=21,header= None,sep=r'\s+')
# print(df)
df.columns =['DATE','TIME','DOY','H','D','Z','F']
# print(df)
df['DATETIME'] = df['DATE'] + ':' + df['TIME']
# print(df['DATETIME'])
df = df[['DATETIME', 'H','D','Z','F']]
print(df)
df['DATETIME'] = pd.to_datetime(df['DATETIME'],format= '%Y-%m-%d:%H:%M:%S.%f')
# print(df['DATETIME'])
```

Plotting H component(horizontal magnetic field), D component(magnetic declination variations),<br>       Z component(vertical magnetic field), F(total magentic field intensity) for  required 5 days

```python
fig,ax = plt.subplots(4,1,figsize=(30,20),sharex= True,layout='constrained')

plt.suptitle('Alibag Magnetic Field Response to the January 2026 Geomagnetic Storm',fontsize= 20)

ax[0].plot(df['DATETIME'],df['H'])
my_fmt = mdates.DateFormatter('%d %b')
ax[0].xaxis.set_major_formatter(my_fmt)
ax[0].xaxis.set_major_locator(mdates.DayLocator(interval=1))
ax[0].xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))
ax[0].set_xlim(np.min(df['DATETIME']), np.max(df['DATETIME']))
ax[0].grid(True)
ax[0].set_ylabel('H(nT)',size= 15)

ax[1].plot(df['DATETIME'],df['D'],c='green')
my_fmt = mdates.DateFormatter('%d %b')
ax[1].xaxis.set_major_formatter(my_fmt)
ax[1].set_xlim(np.min(df['DATETIME']), np.max(df['DATETIME']))
ax[1].grid(True)
ax[1].set_ylabel('D(nT)',size= 15)

ax[2].plot(df['DATETIME'],df['Z'],c='red')
my_fmt = mdates.DateFormatter('%d %b')
ax[2].xaxis.set_major_formatter(my_fmt)
ax[2].set_xlim(np.min(df['DATETIME']), np.max(df['DATETIME']))
ax[2].grid(True)
ax[2].set_ylabel('Z(nT)',size= 15)

ax[3].plot(df['DATETIME'],df['F'],c='orange')
my_fmt = mdates.DateFormatter('%d %b')
ax[3].xaxis.set_major_formatter(my_fmt)
ax[3].set_xlim(np.min(df['DATETIME']), np.max(df['DATETIME']))
ax[3].grid(True)
ax[3].set_xlabel('Date(UT)',size= 15)
ax[3].set_ylabel('F(nT)',size= 15)
# plt.savefig('Alibag_Mag_Ground_Response.pdf',dpi=1000)
plt.savefig("Alibag_Mag_Ground_Response.png", dpi=300)
plt.show()
```

- Between 19 to 20 January 2026, 
    * H component reaches minimum shortly after the strom beginning.
    * D component shows strong fluctuations around storm onset.
    * Z component displayed increased variation.
    * F component shows sharp disturbance.
