# Ionospheric Response  

how ionosphere reacts to external energy inputs like solar flares,geomagnetic strom and meteorological waves.

<b>Ionosphere</b>
- Present in Thermosphere
- Created by solar radiation
- Reason for aurora and helps reflecting radiowaves
- It has 3 layers:
    * D layer --> only exist during day
    * E layer --> it continues to hold a small amount of change at night
    * F layer --> highest concentration of ions and remains charged even at night

```python
import cdflib              # for cdf file
import matplotlib.pyplot as plt
import numpy as np
```

reading the file

```python
cdf1 = cdflib.CDF(r'ion_16_jan_2026.cdf')
cdf2 = cdflib.CDF(r'ion_17_jan_2026.cdf')
cdf3 = cdflib.CDF(r'ion_18_jan_2026.cdf')
cdf4 = cdflib.CDF(r'ion_19_jan_2026.cdf')
cdf5 = cdflib.CDF(r'ion_20_jan_2026.cdf')
cdf6 = cdflib.CDF(r'ion_21_jan_2026.cdf')
cdf7 = cdflib.CDF(r'ion_22_jan_2026.cdf')
cdf8 = cdflib.CDF(r'ion_23_jan_2026.cdf')
cdf9 = cdflib.CDF(r'ion_24_jan_2026.cdf')
cdf10 = cdflib.CDF(r'ion_25_jan_2026.cdf')
cdf11 = cdflib.CDF(r'ion_26_jan_2026.cdf')
```

getting all variables in a single cdf file

```python
for var in cdf2.cdf_info().zVariables:
       print(var)
```

extract the variable of time and converted to datetime (returns in numpy)

```python
time1 = cdf1.varget('Timestamp')
time_dt1 = cdflib.cdfepoch.to_datetime(time1)
time2 = cdf2.varget('Timestamp')
time_dt2 = cdflib.cdfepoch.to_datetime(time2)
time3 = cdf3.varget('Timestamp')
time_dt3 = cdflib.cdfepoch.to_datetime(time3)
time4 = cdf4.varget('Timestamp')
time_dt4 = cdflib.cdfepoch.to_datetime(time4)
time5 = cdf5.varget('Timestamp')
time_dt5 = cdflib.cdfepoch.to_datetime(time5)
time6 = cdf6.varget('Timestamp')
time_dt6 = cdflib.cdfepoch.to_datetime(time6)
time7 = cdf7.varget('Timestamp')
time_dt7 = cdflib.cdfepoch.to_datetime(time7)
time8 = cdf8.varget('Timestamp')
time_dt8 = cdflib.cdfepoch.to_datetime(time8)
time9 = cdf9.varget('Timestamp')
time_dt9 = cdflib.cdfepoch.to_datetime(time9)
time10 = cdf10.varget('Timestamp')
time_dt10 = cdflib.cdfepoch.to_datetime(time10)
time11 = cdf11.varget('Timestamp')
time_dt11 = cdflib.cdfepoch.to_datetime(time11)
# print(time_dt1)
```

Combined all datetime using numpy

```python
combined_time_dt = np.concatenate([time_dt1 , time_dt2 , time_dt3 , time_dt4 , time_dt5 , time_dt6 , time_dt7 , time_dt8 , time_dt9 , time_dt10 , time_dt11])

print(combined_time_dt)
```

extract the variable of ion density (returns in numpy)

```python
ion_density1 = cdf1.varget('N_ion')
ion_density2 = cdf2.varget('N_ion')
ion_density3 = cdf3.varget('N_ion')
ion_density4 = cdf4.varget('N_ion')
ion_density5 = cdf5.varget('N_ion')
ion_density6 = cdf6.varget('N_ion')
ion_density7 = cdf7.varget('N_ion')
ion_density8 = cdf8.varget('N_ion')
ion_density9 = cdf9.varget('N_ion')
ion_density10 = cdf10.varget('N_ion')
ion_density11 = cdf11.varget('N_ion')

print(ion_density1)
```

Combined all ion density using numpy

```python
combined_ion = np.concatenate([ion_density1,ion_density2,ion_density3,ion_density4,ion_density5,ion_density6,ion_density7,ion_density8,ion_density9,ion_density10,ion_density11])
print(combined_ion)
```

Plotting the ion density and date

```python
plt.figure(figsize=(30,5))
plt.plot(combined_time_dt,combined_ion,linewidth = 0.5)

plt.title('Swarm-Ion Density Plot (16-26 January 2026)',fontsize = 40)
plt.xlim(combined_time_dt.min(),combined_time_dt.max())
plt.xlabel('Date',fontsize = 20)
plt.ylabel('Ion Density',fontsize = 20)
# plt.savefig('Ion_Density_Response.pdf',dpi=1000)
plt.savefig("Ion_Density_Response.png", dpi=300)
plt.show()
```
