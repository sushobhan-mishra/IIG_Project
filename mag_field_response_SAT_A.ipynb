# magnetic field response  (Swarm SAT A) 

```python
import cdflib              # cdf files
import matplotlib.pyplot as plt
import numpy as np
```

reading the file

```python
cdf1 = cdflib.CDF(r'mag_SAT_A_16_jan_2026.cdf')
cdf2 = cdflib.CDF(r'mag_SAT_A_17_jan_2026.cdf')
cdf3 = cdflib.CDF(r'mag_SAT_A_18_jan_2026.cdf')
cdf4 = cdflib.CDF(r'mag_SAT_A_19_jan_2026.cdf')
cdf5 = cdflib.CDF(r'mag_SAT_A_20_jan_2026.cdf')
cdf6 = cdflib.CDF(r'mag_SAT_A_21_jan_2026.cdf')
cdf7 = cdflib.CDF(r'mag_SAT_A_22_jan_2026.cdf')
cdf8 = cdflib.CDF(r'mag_SAT_A_23_jan_2026.cdf')
cdf9 = cdflib.CDF(r'mag_SAT_A_24_jan_2026.cdf')
cdf10 = cdflib.CDF(r'mag_SAT_A_25_jan_2026.cdf')
cdf11 = cdflib.CDF(r'mag_SAT_A_26_jan_2026.cdf')
# print(cdf1)
```

getting all variables in a single cdf file

```python
for var in cdf1.cdf_info().zVariables:
       print(var)
```

extract the variable of time and converted to datetime (returns in numpy)

```python
time1 = cdf1.varget('Timestamp')
time_dt1 = cdflib.cdfepoch.to_datetime(time1)
time2 = cdf2.varget('Timestamp')
time_dt2 = cdflib.cdfepoch.to_datetime(time2)
time3 = cdf3.varget('Timestamp')
time_dt3 = cdflib.cdfepoch.to_datetime(time3)
time4 = cdf4.varget('Timestamp')
time_dt4 = cdflib.cdfepoch.to_datetime(time4)
time5 = cdf5.varget('Timestamp')
time_dt5 = cdflib.cdfepoch.to_datetime(time5)
time6 = cdf6.varget('Timestamp')
time_dt6 = cdflib.cdfepoch.to_datetime(time6)
time7 = cdf7.varget('Timestamp')
time_dt7 = cdflib.cdfepoch.to_datetime(time7)
time8 = cdf8.varget('Timestamp')
time_dt8 = cdflib.cdfepoch.to_datetime(time8)
time9 = cdf9.varget('Timestamp')
time_dt9 = cdflib.cdfepoch.to_datetime(time9)
time10 = cdf10.varget('Timestamp')
time_dt10 = cdflib.cdfepoch.to_datetime(time10)
time11 = cdf11.varget('Timestamp')
time_dt11 = cdflib.cdfepoch.to_datetime(time11)
# print(time_dt1)
```

Combined all datetime using numpy

```python
combined_time_dt = np.concatenate([time_dt1 , time_dt2 , time_dt3 , time_dt4 , time_dt5 , time_dt6 , time_dt7 , time_dt8 , time_dt9 , time_dt10 , time_dt11])

print(combined_time_dt)
```

Extract Magnetic Field Components (B_NEC)
Extract the North, East, and Center components from each CDF file

```python
B1 = cdf1.varget('B_NEC')
N1 = B1[:,0]
E1 = B1[:,1]
C1 = B1[:,2]

B2 = cdf2.varget('B_NEC')
N2 = B2[:,0]
E2 = B2[:,1]
C2 = B2[:,2]

B3 = cdf3.varget('B_NEC')
N3 = B3[:,0]
E3 = B3[:,1]
C3 = B3[:,2]

B4 = cdf4.varget('B_NEC')
N4 = B4[:,0]
E4 = B4[:,1]
C4 = B4[:,2]

B5 = cdf5.varget('B_NEC')
N5 = B5[:,0]
E5 = B5[:,1]
C5 = B5[:,2]

B6 = cdf6.varget('B_NEC')
N6 = B6[:,0]
E6 = B6[:,1]
C6 = B6[:,2]

B7 = cdf7.varget('B_NEC')
N7 = B7[:,0]
E7 = B7[:,1]
C7 = B7[:,2]

B8 = cdf8.varget('B_NEC')
N8 = B8[:,0]
E8 = B8[:,1]
C8 = B8[:,2]

B9 = cdf9.varget('B_NEC')
N9 = B9[:,0]
E9 = B9[:,1]
C9 = B9[:,2]

B10 = cdf10.varget('B_NEC')
N10 = B10[:,0]
E10 = B10[:,1]
C10 = B10[:,2]

B11 = cdf11.varget('B_NEC')
N11 = B11[:,0]
E11 = B11[:,1]
C11 = B11[:,2]
# print(N11)
```

```python
combined_N = np.concatenate([N1, N2, N3, N4, N5, N6, N7, N8, N9, N10, N11])
combined_E = np.concatenate([E1, E2, E3, E4, E5, E6, E7, E8, E9, E10, E11])
combined_C = np.concatenate([C1, C2, C3, C4, C5, C6, C7, C8, C9, C10, C11])

print(combined_N)
print(combined_E)
print(combined_C)
```

Combine all magnetic field components across all days

```python
F1 = cdf1.varget('F')
F2 = cdf2.varget('F')
F3 = cdf3.varget('F')
F4 = cdf4.varget('F')
F5 = cdf5.varget('F')
F6 = cdf6.varget('F')
F7 = cdf7.varget('F')
F8 = cdf8.varget('F')
F9 = cdf9.varget('F')
F10 = cdf10.varget('F')
F11 = cdf11.varget('F')

# print(F1)
```

Extract Total Magnetic Field Magnitude (F)

```python
combined_F = np.concatenate([F1, F2, F3, F4, F5, F6, F7, F8, F9, F10, F11])
print(combined_F)
```

```python
lat1 = cdf1.varget('Latitude')
lat2 = cdf2.varget('Latitude')
lat3 = cdf3.varget('Latitude')
lat4 = cdf4.varget('Latitude')
lat5 = cdf5.varget('Latitude')
lat6 = cdf6.varget('Latitude')
lat7 = cdf7.varget('Latitude')
lat8 = cdf8.varget('Latitude')
lat9 = cdf9.varget('Latitude')
lat10 = cdf10.varget('Latitude')
lat11 = cdf11.varget('Latitude')
```

```python
combined_lat = np.concatenate([lat1, lat2, lat3, lat4, lat5, lat6, lat7, lat8, lat9, lat10, lat11])
print(combined_lat)
```

```python
lon1 = cdf1.varget('Longitude')
lon2 = cdf2.varget('Longitude')
lon3 = cdf3.varget('Longitude')
lon4 = cdf4.varget('Longitude')
lon5 = cdf5.varget('Longitude')
lon6 = cdf6.varget('Longitude')
lon7 = cdf7.varget('Longitude')
lon8 = cdf8.varget('Longitude')
lon9 = cdf9.varget('Longitude')
lon10 = cdf10.varget('Longitude')
lon11 = cdf11.varget('Longitude')
```

```python
combined_lon = np.concatenate([lon1, lon2, lon3, lon4, lon5, lon6, lon7, lon8, lon9, lon10, lon11])
print(combined_lon)
```

Plot the Magnetic Field Response

```python
fig,ax = plt.subplots(4,1,figsize=(100,30),sharex=True,constrained_layout=True)

ax[0].plot(combined_time_dt, combined_N)
ax[0].grid(True)
ax[0].set_ylabel('N (nT)',fontsize = 50)

ax[1].plot(combined_time_dt, combined_E)
ax[1].grid(True)
ax[1].set_ylabel('E (nT)',fontsize = 50)

ax[2].plot(combined_time_dt, combined_C)
ax[2].grid(True)
ax[2].set_ylabel('C (nT)',fontsize = 50)

ax[3].plot(combined_time_dt, combined_F)
ax[3].grid(True)
ax[3].set_xlim(combined_time_dt.min(), combined_time_dt.max())
ax[3].set_ylabel('F (nT)',fontsize = 50)
ax[3].set_xlabel('Date',fontsize = 50)

plt.suptitle('Swarm-A Magnetic Field Response',fontsize=200)
# plt.savefig('SAT_A_Magnetic_Response.pdf',dpi=1000)
# plt.savefig("SAT_A_Magnetic_Response.png", dpi=300)
plt.show()
```
